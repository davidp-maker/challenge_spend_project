```python
import pandas as pd, requests, io, numpy as np

BASE   = '/content/drive/MyDrive/challenge_spend_project/data/'
stores = pd.read_csv(BASE + 'sweetgreen_all_us_locations.csv')
aff    = pd.read_csv(BASE + 'aff_addy.csv')

# address -> lat/lon via the free US Census geocoder
def geocode(df, id_col, street, city, state, zip_):
    b = df[[id_col, street, city, state, zip_]].fillna('')
    r = requests.post(
        'https://geocoding.geo.census.gov/geocoder/locations/addressbatch',
        files={'addressFile': ('a.csv', b.to_csv(index=False, header=False))},
        data={'benchmark': 'Public_AR_Current'}, timeout=300)
    cols = ['id','input','match','mtype','matched','lonlat','tiger','side']
    res = pd.read_csv(io.StringIO(r.text), names=cols)
    res[['lon','lat']] = res['lonlat'].str.split(',', expand=True)
    res['lat'] = pd.to_numeric(res['lat'], errors='coerce')
    res['lon'] = pd.to_numeric(res['lon'], errors='coerce')
    return res.set_index('id')[['lat','lon']]

stores = stores.reset_index()
stores = stores.join(geocode(stores,'index','Street Address','City','State','Zip'), on='index').dropna(subset=['lat','lon'])
print(len(stores), 'stores placed on the map')

aff = aff.join(geocode(aff,'affiliate_id','address1','city','state','zip'), on='affiliate_id')
print(aff['lat'].notna().sum(), 'creators placed on the map')

def hav(a1,o1,a2,o2):
    R=3958.8; p1,p2=np.radians(a1),np.radians(a2)
    dp,dl=np.radians(a2-a1),np.radians(o2-o1)
    x=np.sin(dp/2)**2+np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2*R*np.arcsin(np.sqrt(x))

D = hav(aff['lat'].values[:,None], aff['lon'].values[:,None],
        stores['lat'].values[None,:], stores['lon'].values[None,:])
aff['miles_to_nearest_sg'] = np.nanmin(D, axis=1)
print(aff['miles_to_nearest_sg'].describe())
```

```python
# Cell 3 — features + resolved set
DROP = ['realized_payout','payout_log','payout_status','enrolled_at','affiliate_id','challenge_id','month']
FEATURES = [c for c in df.columns if c not in DROP]

# safety check: every feature must be numeric now
bad = [c for c in FEATURES if df[c].dtype == 'object']
print('non-numeric features (should be empty):', bad)

resolved = df[df.payout_status.isin(['completed','skipped'])].copy()
```

```python
rows = []
for m in sorted(resolved.month.unique()):
    train = resolved[resolved.month < m]
    test  = resolved[resolved.month == m]
    if len(train) < 200 or test.realized_payout.sum() == 0:
        continue
    pred, act = hurdle_cal(train, test).sum(), test.realized_payout.sum()
    rows.append((m, round(pred), round(act), round((pred-act)/act*100, 1)))

bt = pd.DataFrame(rows, columns=['month','pred','actual','err_%'])
print(bt)
```

```python
# Improvement 1: calibrated hurdle (fixes the under-bias)
from sklearn.calibration import CalibratedClassifierCV

def hurdle_cal(train, test):
    base = HistGradientBoostingClassifier(random_state=42, min_samples_leaf=50,
               l2_regularization=1, learning_rate=0.1)
    clf = CalibratedClassifierCV(base, method='sigmoid', cv=3).fit(
               train[FEATURES], (train.realized_payout>0).astype(int))
    ea  = train[train.realized_payout > 0]
    reg = HistGradientBoostingRegressor(random_state=42, min_samples_leaf=50,
               l2_regularization=1).fit(ea[FEATURES], ea.payout_log)
    smear = np.mean(np.exp(ea.payout_log - reg.predict(ea[FEATURES])))
    p   = clf.predict_proba(test[FEATURES])[:, 1]
    amt = np.minimum(np.expm1(reg.predict(test[FEATURES]))*smear,
                     test['max_cpm_reward'].fillna(1e9).values)
    return p * amt
```

```python
# Improvement 2: single Tweedie model (the insurance-claims standard for $ like this)
!pip install lightgbm -q
from lightgbm import LGBMRegressor

def tweedie(train, test):
    m = LGBMRegressor(objective='tweedie', tweedie_variance_power=1.3,
                      n_estimators=300, learning_rate=0.05, num_leaves=31,
                      min_child_samples=50, reg_lambda=1, random_state=42, verbose=-1)
    m.fit(train[FEATURES], train.realized_payout)      # predicts $ directly, zeros and all
    return np.clip(m.predict(test[FEATURES]), 0, None)
```

```python
# compare all three, same walk-forward
def backtest(fn):
    out = []
    for mth in sorted(resolved.month.unique()):
        tr, te = resolved[resolved.month < mth], resolved[resolved.month == mth]
        if len(tr) < 200 or te.realized_payout.sum() == 0: continue
        out.append((mth, round(fn(tr, te).sum()), round(te.realized_payout.sum())))
    r = pd.DataFrame(out, columns=['month','pred','actual'])
    r['err_%'] = ((r.pred - r.actual)/r.actual*100).round(1)
    return r

print('Calibrated hurdle:\n', backtest(hurdle_cal), '\n')
print('Tweedie:\n',           backtest(tweedie))
```

```python
from sklearn.inspection import permutation_importance

# fit the P(earn) classifier once on all resolved history
base = HistGradientBoostingClassifier(random_state=42, min_samples_leaf=50,
                                      l2_regularization=1, learning_rate=0.1)
clf = CalibratedClassifierCV(base, method='sigmoid', cv=3).fit(
    resolved[FEATURES], (resolved.realized_payout > 0).astype(int))

# permutation importance: how much AUC drops when each feature is shuffled
imp = permutation_importance(clf, resolved[FEATURES],
                             (resolved.realized_payout > 0).astype(int),
                             scoring='roc_auc', n_repeats=10, random_state=42)

fi = (pd.Series(imp.importances_mean, index=FEATURES)
        .sort_values(ascending=False))
print(fi.round(4))

fi.head(12).sort_values().plot.barh(figsize=(8,6))
plt.title('Feature importance (drop in AUC when shuffled)')
plt.xlabel('importance'); plt.tight_layout(); plt.show()
```
