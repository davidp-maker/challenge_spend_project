```python
import pandas as pd, requests, io, numpy as np

BASE   = '/content/drive/MyDrive/challenge_spend_project/data/'
stores = pd.read_csv(BASE + 'sweetgreen_all_us_locations.csv')
aff    = pd.read_csv(BASE + 'aff_addy.csv')

# address -> lat/lon via the free US Census geocoder
def geocode(df, id_col, street, city, state, zip_):
    b = df[[id_col, street, city, state, zip_]].fillna('')
    r = requests.post(
        'https://geocoding.geo.census.gov/geocoder/locations/addressbatch',
        files={'addressFile': ('a.csv', b.to_csv(index=False, header=False))},
        data={'benchmark': 'Public_AR_Current'}, timeout=300)
    cols = ['id','input','match','mtype','matched','lonlat','tiger','side']
    res = pd.read_csv(io.StringIO(r.text), names=cols)
    res[['lon','lat']] = res['lonlat'].str.split(',', expand=True)
    res['lat'] = pd.to_numeric(res['lat'], errors='coerce')
    res['lon'] = pd.to_numeric(res['lon'], errors='coerce')
    return res.set_index('id')[['lat','lon']]

stores = stores.reset_index()
stores = stores.join(geocode(stores,'index','Street Address','City','State','Zip'), on='index').dropna(subset=['lat','lon'])
print(len(stores), 'stores placed on the map')

aff = aff.join(geocode(aff,'affiliate_id','address1','city','state','zip'), on='affiliate_id')
print(aff['lat'].notna().sum(), 'creators placed on the map')

def hav(a1,o1,a2,o2):
    R=3958.8; p1,p2=np.radians(a1),np.radians(a2)
    dp,dl=np.radians(a2-a1),np.radians(o2-o1)
    x=np.sin(dp/2)**2+np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2*R*np.arcsin(np.sqrt(x))

D = hav(aff['lat'].values[:,None], aff['lon'].values[:,None],
        stores['lat'].values[None,:], stores['lon'].values[None,:])
aff['miles_to_nearest_sg'] = np.nanmin(D, axis=1)
print(aff['miles_to_nearest_sg'].describe())
```

```python
!pip install pgeocode -q
import pgeocode
nomi = pgeocode.Nominatim('us')

miss = aff['lat'].isna()
z = aff.loc[miss, 'zip'].astype(str).str.extract(r'(\d{5})')[0]
g = nomi.query_postal_code(z.tolist())
aff.loc[miss, 'lat'] = g['latitude'].values
aff.loc[miss, 'lon'] = g['longitude'].values

print('still missing after zip fallback:', aff['lat'].isna().sum())
```

```python
!pip install geopy -q
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geo = RateLimiter(Nominatim(user_agent='challenge_spend').geocode, min_delay_seconds=1)

for i in aff[aff['lat'].isna()].index:
    loc = geo(f"{aff.at[i,'city']}, {aff.at[i,'state']}, USA")
    if loc:
        aff.at[i, 'lat'], aff.at[i, 'lon'] = loc.latitude, loc.longitude

print('still missing:', aff['lat'].isna().sum())
```

```python
# 1. miles feature onto df
df = df.merge(aff[['affiliate_id','miles_to_nearest_sg']], on='affiliate_id', how='left')

# 2. participation, ordered by time so it's non-leaky
df = df.sort_values('start_date').reset_index(drop=True)
df['prior_completions'] = (df.groupby('affiliate_id')['completed']
                             .transform(lambda s: s.shift(fill_value=False).cumsum()))

# analysis subset
r = df[df.payout_status.isin(['completed','skipped'])].copy()
r['earned'] = (r.realized_payout > 0).astype(int)

# 3a. distance vs outcome  (your main hypothesis)
bins   = [0,1,3,5,10,25,100,1e9]
labels = ['0-1','1-3','3-5','5-10','10-25','25-100','100+']
r['dist_bucket'] = pd.cut(r['miles_to_nearest_sg'], bins=bins, labels=labels, right=False)
print(r.groupby('dist_bucket').agg(
    creators=('affiliate_id','size'), earn_rate=('earned','mean'),
    total_spend=('realized_payout','sum')).round(3))

print()
# 3b. participation vs outcome
print(r.groupby('prior_completions').agg(
    rows=('earned','size'), earn_rate=('earned','mean')).round(3))
```

```python
import seaborn as sns, matplotlib.pyplot as plt

r = df[df.payout_status.isin(['completed','skipped'])].copy()
r['proximity'] = pd.cut(r['miles_to_nearest_sg'], [0,5,25,1e9],
                        labels=['near <5mi','mid 5-25','far 25+'])

# 1. P(payout | proximity): payout density among earners, split by distance
e = r[r.realized_payout > 0]
plt.figure(figsize=(9,5))
sns.kdeplot(data=e, x='realized_payout', hue='proximity',
            common_norm=False, fill=True, alpha=0.25, clip=(0,800))
plt.title('P(payout | proximity) — earners only')
plt.show()

# 2. distance density: creators vs where the dollars are
plt.figure(figsize=(9,5))
sns.kdeplot(x=r['miles_to_nearest_sg'], clip=(0,60), fill=True, alpha=0.2, label='creators')
sns.kdeplot(x=r['miles_to_nearest_sg'], weights=r['realized_payout'], clip=(0,60),
            fill=True, alpha=0.2, label='dollars')
plt.title('Distance density: creators vs dollars')
plt.xlabel('miles to nearest Sweetgreen'); plt.legend(); plt.show()
```
